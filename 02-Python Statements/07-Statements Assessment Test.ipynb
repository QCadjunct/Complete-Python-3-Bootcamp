**Use <code>for</code>, .split(), and <code>if</code> to create a Statement that will print out words that start with 's':**

```python
st = 'Print only the words that start with s in this sentence'
```

```python
#Code here
sentence= st.split()
word_starts_with_s = []
for word in sentence :
    if (word[0] == 's') : word_starts_with_s.append(word)
print (word_starts_with_s)
```

```python
#Code here
sentence= st.split()
word_contains_the_letter_s = []
for word in sentence :
    if ('s' in word) : word_contains_the_letter_s.append(word)
print (word_contains_the_letter_s)
```

**Write a program that prints the integers from 1 to 100. But for multiples of three print "Fizz" instead of the number, and for the multiples of five print "Buzz". For numbers which are multiples of both three and five print "FizzBuzz".**

```python
for idx,num in enumerate(range(1,101)):
    if (num % 3 == 0 and num % 5 == 0 ) : print (f'Index: {idx} : Value: {num} divisible by 3 and 5 so FizzBuzz')
    elif (num % 3 == 0 ) : print (f'Index: {idx} : Value: {num} divisible by 3 so Fizz')
    elif (num % 5 == 0 ) : print (f'Index: {idx} : Value: {num} divisible by 5 so Buzz')
    else : print (f'Index: {idx} : Value: {num} it is not divisible by 3 or 5 so ???????')
```
