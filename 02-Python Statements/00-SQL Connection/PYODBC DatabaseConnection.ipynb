```python
import pyodbc

class DatabaseConnection:
    def __init__(self):
        self.connection = None

    def create_pyodbc_connection(self, connection_string):
        self.connection = pyodbc.connect(connection_string)
        return self

    def execute_sql(self, sql_command):
        with self.connection.cursor() as cursor:
            cursor.execute(sql_command)
            # Assuming a fetch operation for simplicity
            return cursor.fetchall()

# Usage
db = DatabaseConnection()
results = db.create_pyodbc_connection("your_connection_string").execute_sql("SELECT * FROM your_table")
print(results)
```
