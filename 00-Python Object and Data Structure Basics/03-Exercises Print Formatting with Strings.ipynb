```python
# Indexed by ordinal poistion                      0            1
print('String here {1} then also {0}'.format('something1','something2'))
```

```python
# Assigning print paramter names in the format FOX='fox',QUICK='quick',BROWN='brown' and use within the print 'The {QUICK} {BROWN} {FOX}'
print('The {QUICK} {BROWN} {FOX}'.format(FOX='fox',QUICK='quick',BROWN='brown'))
```

```python
result = 100/777
```

```python
result
```

```python
# {RESULT:1.5f}, the f is for float formatting
print('The result was {RESULT:1.5f}'.format(RESULT=result))
```

```python
# f string replaces the .format similar to macro substitution in erwin
FOX='fox'
QUICK='quick'
BROWN='brown'
print(f'The {QUICK} {BROWN} {FOX}')
```
